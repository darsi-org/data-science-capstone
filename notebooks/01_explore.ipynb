```python
import os
import subprocess
import sys

REPO = "https://github.com/<your-username>/data-science-capstone"  # Colab only: your copy's address

if "google.colab" in sys.modules and not os.path.exists("check.py"):
    if not os.path.isdir("data-science-capstone"):
        subprocess.run(["git", "clone", REPO], check=True)
    os.chdir("data-science-capstone")
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")  # VS Code and Jupyter start in notebooks/
if not os.path.exists("data/students.csv"):
    if "google.colab" in sys.modules:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ucimlrepo"], check=True)
    subprocess.run([sys.executable, "fetch_data.py"], check=True)
os.makedirs("figures", exist_ok=True)
print("Working in", os.getcwd())
```
